14. Write a program to reverse a string using a while loop.

```python
# Reverse a string using a while loop

# Get user input
text = input("Enter a string: ")

# Initialize variables
reversed_text = ""
index = len(text) - 1

# Reverse the string using a while loop
while index >= 0:
    reversed_text += text[index]
    index -= 1

# Display the result
print(f"Reversed string: {reversed_text}")
```

15. Write a Python program to calculate the factorial of a number provided by the user using a while loop

```python
# Factorial using a while loop

# Get user input
num = int(input("Enter a non-negative integer: "))

# Check for valid input
if num < 0:
    print("Factorial is not defined for negative numbers.")
else:
  factorial = 1
  n = num

  while n > 1:
    factorial *= n
    n -= 1

  print(f"The factorial of {num} is {factorial}")
```

Type casting, also known as type conversion, is the process of changing a variable's data type into another data type. In Python, this is done using built-in functions.

The user manually converts a variable's data type using built-in functions. These functions include:

10. What is the difference between implicit and explicit type casting?

Implicit type casting happens automatically by the compiler, while explicit type casting requires the programmer to define the conversion using specific syntax.
**Implicit Type Casting (Widening)**:
1. **Example**: Converting an integer (e.g., int) to a floating-point number (e.g., float).
2. ** Purpose**: Ensures compatibility when a smaller data type is assigned to a larger one or when different data types are combined in an expression.
3. ** Safety**: Generally safe as it doesn't lead to loss of data.
**Explicit Type Casting (Narrowing):**
1. **Example:** Converting a floating-point number to an integer.
2. **Purpose**: Provides control over the conversion process and can be used to prevent potential data loss or unexpected behavior.
3. **Syntax**: Typically requires a cast operator (e.g., (int)value).
4. **Safety**: Can lead to data loss or precision issues, especially when converting from a larger floating-point type to an integer.

13. What is the difference between for and while loops?

The main difference between for and while loops lies in when the number of iterations is known beforehand. for loops are used when the number of iterations is known or can be easily determined at the beginning of the loop, while While loops are used when the number of iterations is dependent on a dynamic condition and is not known in advance.
**Difference are as follows**:
1. **for loop**:
Ideal when you know the exact number of times you need to repeat a block of code.
Often used for iterating over sequences (like lists, strings, or arrays) or when you have a clear starting point, condition, and increment/decrement.
The number of iterations is typically determined by a counter variable or a sequence.
Example: "Print numbers from 1 to 10".
2. **while loop**:
Best used when the number of iterations is unknown and depends on a condition that is checked repeatedly.
The loop continues to execute as long as the condition remains true.
Example: "Keep asking for user input until a valid number is entered".
